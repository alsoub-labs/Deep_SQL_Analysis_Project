## *Phase 3 — GROUP BY*

--------------------------------------
*Which category generates the highest revenue?*

```python
%sql

WITH CTE_RANK AS(
SELECT
  cat.Category_Name,
  sum(fac.Sales) AS Total_Sales,
  rank() OVER (ORDER BY sum(fac.Sales) DESC) AS Total_SalesRank
FROM
bara_slaes_project.gold.factable AS fac
  
    LEFT JOIN bara_slaes_project.gold.dimproduct_info As cat
      ON fac.Product_srgk = cat.Product_srgk 
GROUP BY
  cat.Category_Name
)
SELECT * FROM CTE_RANK WHERE Total_SalesRank = 1
```

--------------------------------------
*Which Country has the most customers?*

```python
%sql
SELECT * FROM (
SELECT
  Country,
  COUNT(Customer_ID) AS Number_of_Customers_Per_Country,
  rank() OVER (ORDER BY COUNT(Customer_ID) DESC) AS Number_of_Customers_Per_CountryRank
FROM
  bara_slaes_project.gold.dimcustomer_info
  WHERE Customer_ID IS NOT NULL
GROUP BY
  Country
ORDER BY Number_of_Customers_Per_Country DESC
)t
WHERE Number_of_Customers_Per_CountryRank =1 ;
```

--------------------------------------
*Which Countries gets the largest number of products?*

```python
%sql
WITH CTE_ProductsPerCountry AS(

SELECT
  cus.Country,
  COUNT(prod.Product_name) AS Number_of_Products_Per_Country,
  rank() OVER (ORDER BY COUNT(prod.Product_name) DESC) AS Number_of_Products_Per_CountryRank
FROM
  bara_slaes_project.gold.dimcustomer_info As cus
    LEFT JOIN bara_slaes_project.gold.dimproduct_info AS Prod
      ON cus.Customer_srgk = Prod.Product_srgk
GROUP BY
  cus.Country
ORDER BY
  Number_of_Products_Per_Country DESC
)
SELECT * FROM CTE_ProductsPerCountry WHERE Number_of_Products_Per_CountryRank = 1;
```

--------------------------------------
*Average order Price by Country*

```python
%sql
SELECT
  cus.Country,
  fac.Order_Number,
  avg(fac.Price) As Average_Price
FROM
  bara_slaes_project.gold.factable As Fac
    LEFT JOIN bara_slaes_project.gold.dimcustomer_info AS Cus
      ON Fac.Customer_srgk = Cus.Customer_srgk
GROUP BY
  cus.Country,fac.Order_Number;
```

--------------------------------------
*Orders by weekday.*

```python
%sql
SELECT
  dayname(Order_Date) AS Week_Day,
  COUNT(Order_Number) AS Number_of_Orders_By_Week_Day
FROM
  bara_slaes_project.gold.factable
GROUP BY
  Week_Day;
```

--------------------------------------
*Revenue by year.*

```python
%sql
SELECT
  -- year(Order_Date) AS year_number,
  date_format(Order_Date, 'y') AS year_number,
  sum(Sales) AS Total_Sales_By_Year
FROM
  bara_slaes_project.gold.factable
GROUP BY
  year_number;
```

--------------------------------------
*Revenue by month.*

```python
%sql
SELECT
  -- month(Order_Date) AS Month_Name,
  date_format(Order_Date, 'MMM') AS Month_Name,
  sum(Sales) AS Total_Sales_By_Year
FROM
  bara_slaes_project.gold.factable
GROUP BY
  Month_Name;
```

--------------------------------------
*Top five customers by spending.*

```python
%sql
SELECT
  *
FROM
  (
    SELECT
      Cus.Customer_ID,
      concat(Cus.FullName) AS Customer_Name,
      SUM(Fac.Sales) AS TOTAL_SALES,
      Rank() OVER (ORDER BY SUM(Fac.Sales) DESC) AS TOTAL_SALES_Rank
    FROM
      bara_slaes_project.gold.factable AS Fac
        LEFT JOIN bara_slaes_project.gold.dimcustomer_info AS Cus
          ON Fac.Customer_srgk = Cus.Customer_srgk
    WHERE
      Cus.Customer_ID IS NOT NULL
    GROUP BY
      Cus.Customer_ID,
      Customer_Name
  ) t
WHERE
  total_sales_rank IN (1, 2, 3, 4, 5)
```

--------------------------------------
*Bottom five customers.*

```python
%sql
SELECT
  *
FROM
  (
    SELECT
      Cus.Customer_ID,
      concat(Cus.FullName) AS Customer_Name,
      SUM(Fac.Sales) AS TOTAL_SALES,
      dense_rank() OVER (ORDER BY SUM(Fac.Sales) ASC) AS TOTAL_SALES_Rank
    FROM
      bara_slaes_project.gold.factable AS Fac
        LEFT JOIN bara_slaes_project.gold.dimcustomer_info AS Cus
          ON Fac.Customer_srgk = Cus.Customer_srgk
    WHERE
      Cus.Customer_ID IS NOT NULL
    GROUP BY
      Cus.Customer_ID,
      Customer_Name
  ) t
WHERE
  total_sales_rank IN (1, 2, 3, 4, 5)
```
