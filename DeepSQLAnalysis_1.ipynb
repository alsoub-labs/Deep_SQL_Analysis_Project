## *Phase 1 — Basic Queries (Foundation)*

*Display every customer's full name in uppercase.*

```python
%sql
SELECT
  upper(FullName) AS CustomerFullNameUppercase
FROM
  bara_slaes_project.gold.dimcustomer_info
LIMIT 5;
```

*Display the first three letters of each product.*

```python
%sql
SELECT
  trim(left(Product_name, 3)) AS Product_1st3_chars
FROM
  bara_slaes_project.gold.dimproduct_info
LIMIT 5;
```

*Find products whose names contain the word "Road".*

```python
%sql
SELECT
  Product_name
FROM
  bara_slaes_project.gold.dimproduct_info
WHERE
Product_name LIKE '%Road%'
LIMIT 5;
```

*Remove extra spaces from customer names.*

```python
%sql
SELECT trim(FullName) FROM bara_slaes_project.gold.dimcustomer_info LIMIT 5;
```

*Display product names with their character length.*

```python
%sql
SELECT
  Product_name,
  len(Product_name) As ProductLength
FROM
  bara_slaes_project.gold.dimproduct_info
LIMIT 5;
```

*Round every product price to the nearest integer.*

```python
%sql
SELECT DISTINCT
  (Prod.Product_name),
  Fact.Price,
  INT(Fact.Price) AS priceinteger
FROM
  bara_slaes_project.gold.factable AS Fact
    LEFT JOIN bara_slaes_project.gold.dimproduct_info AS Prod
      ON Prod.Product_srgk = Fact.Product_srgk;
```

*Display the square root of every product Price.*

```python
%sql
SELECT DISTINCT
  (Prod.Product_name),
  Fact.Price,
  sqrt(Fact.Price) AS sqrt_product_Price
FROM
  bara_slaes_project.gold.factable AS Fact
    LEFT JOIN bara_slaes_project.gold.dimproduct_info AS Prod
      ON Prod.Product_srgk = Fact.Product_srgk;
```

*Show order IDs concatenated with customer names.*

```python
%sql
SELECT
  fact.Order_Number,
  cus.FullName,
  concat(fact.Order_Number,"_",cus.FullName) AS Cusomer_with_orderNumber
FROM
  bara_slaes_project.gold.factable AS Fact
  LEFT JOIN bara_slaes_project.gold.dimcustomer_info AS Cus
  ON Fact.Customer_srgk = Cus.Customer_srgk;
```

*Display the current date and time beside every order.*

```python
%sql
SELECT
  Order_Number,
  now() AS CurrentDateAndTime,
  date(date_format(now(), "yyyy-MM-dd")) AS CurrentDate
FROM
  bara_slaes_project.gold.factable;
```
