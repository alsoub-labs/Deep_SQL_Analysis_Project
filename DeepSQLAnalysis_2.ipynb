## *Phase 2 — Aggregate Analysis*

*Calculate total sales.*


```python
%sql
SELECT
  SUM(Sales) AS Total_Sales
FROM
  bara_slaes_project.gold.factable;
```

*Find average product price.*

```python
%sql
SELECT
  Prod.Product_name,
  avg(Fac.Price) AS AvgProductPrice
FROM
  bara_slaes_project.gold.factable AS Fac
    left JOIN bara_slaes_project.gold.dimproduct_info AS Prod
      ON Fac.Product_srgk = Prod.Product_srgk
GROUP BY
  Prod.Product_name;
```

*Count total customers.*

```python
%sql
SELECT
  count(DISTINCT Customer_ID) AS Total_Customers
FROM
  bara_slaes_project.gold.dimcustomer_info;
```

*Find the most expensive product.*

```python
%sql
SELECT * FROM (
SELECT 
  DISTINCT  Prod.Product_name As Product,
  FAC.Price As  Price,
  RANK() OVER (ORDER BY FAC.Price DESC) As  PriceRank
FROM
  bara_slaes_project.gold.factable AS Fac
    left JOIN bara_slaes_project.gold.dimproduct_info AS Prod
      ON Fac.Product_srgk = Prod.Product_srgk)
      WHERE PriceRank == 1;
```

*Find the cheapest product.*

```python
%sql
SELECT * FROM (
SELECT 
  DISTINCT  Prod.Product_name As Product,
  FAC.Price As  Price,
  RANK() OVER (ORDER BY FAC.Price ASC) As  PriceRank
FROM
  bara_slaes_project.gold.factable AS Fac
    left JOIN bara_slaes_project.gold.dimproduct_info AS Prod
      ON Fac.Product_srgk = Prod.Product_srgk)
      WHERE PriceRank == 1;
```

*Calculate average order value.*

```python
%sql
SELECT 
  Order_Number,
  avg(Price) AS AvgOrderValue
FROM
  bara_slaes_project.gold.factable
GROUP BY
  Order_Number;
```

*Find total revenue per category.*

```python
%sql
SELECT
  prod.Category_Name,
  SUM(fac.Sales) AS Total_Revenue
FROM
  bara_slaes_project.gold.factable AS Fac
    left JOIN bara_slaes_project.gold.dimproduct_info AS Prod
      ON Fac.Product_srgk = Prod.Product_srgk
GROUP BY
  prod.Category_Name;
```

*Count products per Line.*

```python
%sql
SELECT
  Product_Line,
  count(Product_name) AS Prodcut_Count_Per_Line
FROM
  bara_slaes_project.gold.dimproduct_info
GROUP BY
  Product_Line;
```

*Calculate monthly revenue.*

```python
%sql
SELECT
  date_format(Order_Date, 'MMM') As Month,
  Sum(Sales) AS Monthly_Revenue
FROM
  bara_slaes_project.gold.factable
group by
  Month;
```

*Calculate monthly revenue by year.*

```python
%sql
SELECT
  date_format(Order_Date, 'yyyy') As Year,
  date_format(Order_Date, 'MMM') As Month,
  Sum(Sales) AS Monthly_Revenue_by_Year
FROM
  bara_slaes_project.gold.factable
group by
  Year ,Month;
```

*Calculate average customer spending.*

```python
%sql
SELECT 
  cus.Customer_ID,
  avg(FAC.Sales) AS Avg_Cusomer_Spend
FROM
  bara_slaes_project.gold.factable AS Fac
    left JOIN bara_slaes_project.gold.dimcustomer_info AS cus
      ON Fac.Product_srgk = cus.Customer_srgk
GROUP BY
  cus.Customer_ID;
```
